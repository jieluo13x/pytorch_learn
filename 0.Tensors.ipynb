# cpu vs cuda

**Check if GPU is available**
torch.cuda.is_available()

```python
import time

print("=" * 60)
print("CPU vs GPU ")
print("=" * 60)

# Check if GPU is available
has_cuda = torch.cuda.is_available()
if has_cuda:
    print(f"GPU available: {torch.cuda.get_device_name(0)}")
print()

# ============================================================
# 1. CPU DEVICE (explicit)
# ============================================================
print("1. CPU DEVICE (explicit):")
cpu_tensor = torch.tensor([1, 2, 3],
                          device="cpu")

print(f"   Tensor: {cpu_tensor}")
print(f"   Device: {cpu_tensor.device}")
print()
# ============================================================
# 2. GPU DEVICE (if available)
# ============================================================
if has_cuda:
    print("2. GPU DEVICE:")
    gpu_tensor = torch.tensor([1, 2, 3],
                              device="cuda")
    
    print(f"   Tensor: {gpu_tensor}")
    print(f"   Device: {gpu_tensor.device}")
    
    # Also can use device index (if multiple GPUs)
    gpu_0 = torch.tensor([1, 2, 3],
                         device="cuda:0")  # First GPU
    print(f"   GPU:0 tensor: {gpu_0.device}")
    print()

else:
    print("2. GPU DEVICE: NOT AVAILABLE")

# ============================================================
# 3. Performance comparison (matrix multiplication)
# ============================================================
print("3. PERFORMANCE COMPARISON:")
size = 10000

# CPU computation
cpu_a = torch.randn(size, size, device="cpu")
cpu_b = torch.randn(size, size, device="cpu")

start = time.time()
cpu_result = torch.mm(cpu_a, cpu_b)
cpu_time = time.time() - start
print(f"   CPU time: {cpu_time:.3f} seconds")

if has_cuda:
    # GPU computation
    gpu_a = torch.randn(size, size, device="cuda")
    gpu_b = torch.randn(size, size, device="cuda")
    
    # Warm up GPU
    torch.mm(gpu_a, gpu_b)
    
    start = time.time()
    gpu_result = torch.mm(gpu_a, gpu_b)
    
    gpu_time = time.time() - start
    print(f"   GPU time: {gpu_time:.3f} seconds")
    print(f"   Speedup: {cpu_time / gpu_time:.1f}x faster on GPU")

print("\n✅ Done!")
```
